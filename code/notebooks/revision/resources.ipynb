```python
import pandas as pd
import glob
import os

ct_cols = [
    'Lymphocytes',
    'Neutrophils',
    'Eosinophils',
    'Mast_Cells',
    'Dendritic_Cells',
    'Macrophages',
    'tumor_purity',
]

other_cols = [
    'condition',
    'tissue'
]

tissue_dir = '/home/lnemati/pathway_crosstalk/data/survival_data/tissues'

# All columns to keep
use_cols = ct_cols + other_cols

# Find all CSV files
csv_files = glob.glob(os.path.join(tissue_dir, '*.csv'))

dfs = []
for file in csv_files:
    print(file)
    if file == '/home/lnemati/pathway_crosstalk/data/survival_data/tissues/pan_cancer.csv':
        continue
        
    df = pd.read_csv(
        file,
        usecols=lambda col: col in use_cols or col == 'Unnamed: 0',
        index_col=0
    )
    dfs.append(df)

# Concatenate all dataframes
final_df = pd.concat(dfs)

# Optional: reset index name if needed
# final_df.index.name = None

print(final_df.shape)
```

```python

import pandas as pd
import glob
import os
from concurrent.futures import ProcessPoolExecutor

parent_dir = '/projects/bioinformatics/DB/Xena/TCGA_GTEX/by_tissue_primary_vs_normal'

use_cols = [
    'ligand',
    'receptor',
    'interaction',
    'all_genes',
    'corr'
]

# Use iglob (memory efficient)
csv_iter = glob.iglob(
    os.path.join(parent_dir, '**', 'interactions', 'ccc_lr_pairs.csv'),
    recursive=True
)

def load_file(f):
    parts = f.split(os.sep)
    
    tissue = parts[-5]
    condition = parts[-4]
    cancer_type = parts[-3]
    
    # Try fast engine if available
    try:
        df = pd.read_csv(f, usecols=use_cols, engine="pyarrow")
    except Exception:
        df = pd.read_csv(f, usecols=use_cols)
    
    df['tissue'] = tissue
    df['condition'] = condition
    df['cancer_type'] = cancer_type
    
    return df

# Parallel load
with ProcessPoolExecutor() as executor:
    dfs = list(executor.map(load_file, csv_iter))

# Combine
final_df = pd.concat(dfs, ignore_index=True)

print(f"Loaded {len(dfs)} files, shape: {final_df.shape}")
```

```python
final_df.to_csv('/home/lnemati/pathway_crosstalk/results/corr_interactions/corr_ccc_lr_pairs.csv')
```
